```python
import pandas as pd
import numpy as np
data = {
'ID': range(1, 11),
'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eva',
'Frank', 'Grace', 'Helen', 'Ivan', 'Julia'],
'Age': [23, 35, 29, 40, 28, 45, 32, 27, 38, 31],
'City': ['Moscow', 'SPb', 'Moscow', 'Kazan', 'SPb',
'Moscow', 'SPb', 'Kazan', 'Moscow', 'SPb'],
'Salary': [50000, 70000, 60000, 80000, 55000,
90000, 65000, 58000, 75000, 62000]
}
df = pd.DataFrame(data)
print(df)
print('----------------------------------------')
print('Начало заданий')
print(f'1)колличество строк и столбцов- {df.shape}')

print(f'2)строки 1,3,5,7,9-\n{df.iloc[[0, 2, 4,6,8]]}')

print(f'3)Salary выше медианы-\n{df[df['Salary'] > df['Salary'].median()]}')

print(f'4) Колличество людей в каждом городе-\n{df['City'].value_counts()}')

def rename(City):
    match City:
        case 'Moscow':
            return 'MSK'
        case 'SPb':
            return 'SPB'
        case 'Kazan':
            return 'KZN'


df['City_Code']=df['City'].map(rename)

print(f'добавлен столбец Citu_Code-\n{df}')
```
