```python
import pandas as pd
import numpy as np
# Таблица 1: Продажи
sales = pd.DataFrame({
'OrderID': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
'Product': ['Laptop', 'Phone', 'Tablet', 'Laptop', 'Phone',
'Laptop', 'Tablet', 'Phone', 'Laptop', 'Tablet'],
'Quantity': [1, 2, 1, 3, 1, 2, 1, 4, 1, 2],
'Price': [50000, 30000, 20000, 50000, np.nan,
50000, 20000, 30000, 50000, np.nan],
'Region': ['Moscow', 'SPb', 'Kazan', 'Moscow', 'SPb',
'Kazan', np.nan, 'SPb', 'Moscow', 'Kazan']
})
# Таблица 2: Информация о товарах
products = pd.DataFrame({
'Product': ['Laptop', 'Phone', 'Tablet', 'Monitor'],
'Category': ['Electronics', 'Electronics', 'Electronics', 'Accessories'],
'Weight_kg': [2.5, 0.2, 0.5, 3.0]
})
print("Таблица продаж:")
print(sales)
print("\nТаблица товаров:")
print(products)
print("---------------------------------------------------")
print("начало выполнения задания")
print(f"1) Количество пропусков в столбцах -\n{sales.isnull().sum()}")

print(f'2) средняя цена и количество для каждого продукта-\n{sales.groupby('Product')['Price'].agg(['mean', 'count'])}')

print(f'3) удаление строк с NaN в Price или Region-\n{sales.dropna(subset=['Price', 'Region'])}')

SALES=sales
SALES.columns=sales.columns.str.upper()
print(f'4) столбцы в верхнем регистре-\n{SALES}')

print(f'5) объединение первых 3х строк-\n{pd.concat([sales.head(3), products.head(3)], axis=1)}')
```
